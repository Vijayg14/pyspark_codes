```python
%sql
CREATE TABLE Numbers (
    num INT
);

INSERT INTO Numbers VALUES
(1),
(2),
(3),
(5),
(6),
(8),
(9);
```

```python
%sql

WITH cte AS
(
    SELECT num,
           LEAD(num) OVER(ORDER BY num) AS next_num
    FROM Numbers
)
SELECT num,
       next_num
FROM cte
WHERE next_num - num > 1;
```
